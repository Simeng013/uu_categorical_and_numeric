```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
metabric = pd.read_csv("./data/metabric.csv")
metabric.head()
```

### a.

```python
sns.ecdfplot(
    data=metabric,
    x="Overall Survival (Months)"
)

plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.title("ECDF of Overall Survival")
plt.show()
```

### b.

```python
sns.ecdfplot(
    data=metabric,
    x="Overall Survival (Months)",
    hue="Chemotherapy"
)

plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.title("ECDF of Overall Survival by Chemotherapy")
plt.show()
```

The ECDF for patients who received chemotherapy is generally above and to the left of the ECDF for patients who did not. Therefore, conditional on receiving chemotherapy, we would predict a shorter overall survival time.


### c.

We cannot conclude whether chemotherapy is effective from this graph. Chemotherapy was not randomly assigned, and other variables like the severity of the cancer were not controlled. Patients with more severe cancer may be more likely to receive chemotherapy, while cancer severity itself may also lead to shorter survival. Therefore, the graph just shows association rather than causation.
